```python
import pandas as pd

patients = pd.read_csv("patient.csv")
admissions = pd.read_csv("admission.csv")
diseases = pd.read_csv("disease.csv")

df = pd.merge(admissions, patients, on="patient_id", how="left")
df = pd.merge(df, diseases, on="disease_id", how="left")
```

```python
print("Total admissions by disease")
print(df["disease_name"].value_counts())
```

```python
print("Total admissions by type")
print(df["admission_type"].value_counts())
```

```python
print("admissions cross-tabulation (Disease and admission type)")
print(pd.crosstab(df["disease_name"], df["admission_type"]))
```

```python
print("Total patien per blood group")
print(df["blood_group"].value_counts())
```

```python
print("Total patien per city")
print(df["city"].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

disease_counts = df["disease_name"].value_counts()

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(disease_counts.index, disease_counts.values, color="royalblue", edgecolor="black")
ax.bar_label(bars, label_type="center", color="white", fontweight="bold")
plt.title("Number of Patients per Disease")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="disease_name", hue="gender")
plt.title("Disease Distribution by Gender")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
admission_counts = df["admission_type"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(admission_counts.values, labels=admission_counts.index, autopct="%1.1f%%", wedgeprops={"edgecolor": "black"})
plt.title("Proportion of Admission Types")
plt.show()
```
